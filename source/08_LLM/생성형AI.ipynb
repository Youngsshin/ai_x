```python
from langchain_ollama import ChatOllama
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
llm = ChatOllama(model='llama3.2:1b')
prompt_template = PromptTemplate(
    template = """What is the most famous food in the {country}?
    Return the name of the famous food only""",
    input_variables = ["country"]
)
prompt = prompt_template.invoke({"country":"Korea"})
output_parser = StrOutputParser()
# output_parser.invoke(llm.invoke(chatprompt_template.invoke({"country":"Korea"})))
food_chain = prompt_template | llm | output_parser
food_chain.invoke({"country":"Korea"})
```

```python
from pydantic import BaseModel, Field
recipe_prompt_template = PromptTemplate(
    template = """Give following information about {food}.
    - material
    - recipe
    return it is JSON format and return the JSON dictionary only""",
    input_variables = ["food"]
)
class FoodDetail(BaseModel):
    material:str = Field(description="ingredients of food")
    recipe:str = Field(description="recipe of food")
        
structedllm = llm.with_structured_output(FoodDetail)
info = structedllm.invoke(recipe_prompt_template.invoke({"food":"Pizza"}))
# recipe_chain = recipe_prompt_template | llm | output_parser
# recipe_chain.invoke({"food":"Pizza"})
```

```python
print(info)
```

```python
final_chain = food_chain | recipe_chain
final_chain.invoke({"country":"Hongkong"})
```
