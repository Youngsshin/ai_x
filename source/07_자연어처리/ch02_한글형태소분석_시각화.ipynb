## 3.1 HanNanum

```python
from konlpy.tag import Hannanum
hannanum = Hannanum(jvmpath=None,
                    max_heap_size=1024) # 기본값 : 1024
hannanum.analyze(text)
```

- 구글에서 konlpy docs 검색 후 https://konlpy.org/
- api 메뉴 클릭 -> Comparison between POS tagging classes 클릭 -> Korean POS tags comparison chart 클릭

```python
# 형태소 분석 : morphs
print(hannanum.morphs(text))
```

```python
# 명사만 추출
print(hannanum.nouns(text))
```

```python
# 품사태그
print(hannanum.pos(text, ntags=9)) # 품사의 갯수 기본값 ntags=9
```

```python
print(hannanum.pos(text, ntags=22))
```
